### 4. Загрузка и предварительный анализ всех CSV-файлов

```python
import os
import pandas as pd

# Define data_folder_path here to ensure it's available
data_folder_path = '/content/drive/MyDrive/КурсоваяСинтетическийКонтроль/dataFiles'

# Словарь для хранения всех загруженных DataFrame'ов
dfs = {}

# Получаем список всех файлов в папке
files_in_folder = os.listdir(data_folder_path)

# Фильтруем только CSV-файлы
csv_files = [f for f in files_in_folder if f.endswith('.csv')]

if not csv_files:
    print(f"В папке '{data_folder_path}' не найдено ни одного CSV-файла.")
else:
    print(f"Найдено {len(csv_files)} CSV-файлов. Загружаем...")
    for csv_file_name in csv_files:
        full_csv_path = os.path.join(data_folder_path, csv_file_name)
        print(f"\n--- Загрузка файла: {csv_file_name} ---")
        try:
            # Try reading with default settings first
            df_temp = pd.read_csv(full_csv_path)
        except Exception as e:
            print(f"Произошла ошибка при загрузке или обработке файла {csv_file_name}: {e}")
            # If default fails, try with common alternatives like semicolon delimiter and error handling
            print("Попытка загрузки с другим разделителем (';') или пропуском некорректных строк...")
            try:
                df_temp = pd.read_csv(full_csv_path, sep=';', on_bad_lines='skip') # Added sep=';' and on_bad_lines='skip'
                print("Загружено с альтернативными настройками.")
            except Exception as e_alt:
                print(f"Повторная ошибка при загрузке файла {csv_file_name} с альтернативными настройками: {e_alt}")
                continue # Skip to the next file if even alternative fails

        dfs[csv_file_name] = df_temp # Сохраняем DataFrame в словаре
        print("Файл успешно загружен!")
        print("Первые 5 строк:")
        display(df_temp.head())
        print("Информация о столбцах:")
        df_temp.info()
```

## Melt and Clean GDP Data

### Subtask:
Melt the wide-format GDP data into a long format, converting year columns into a 'Year' column and corresponding values into a 'GDP_Per_Capita' column. Ensure 'Year' is an integer and 'GDP_Per_Capita' is numeric. Rename the state column to 'State'.


**Reasoning**:
Identify the year columns, perform the melt operation, convert data types, and rename the 'GeoName' column to 'State' in a single step.



```python
year_cols = [str(year) for year in range(1970, 2005)]
df_gdp_melted = df_gdp_raw.melt(id_vars=['GeoName'], value_vars=year_cols, var_name='Year', value_name='GDP_Per_Capita')

# Convert 'Year' to integer and 'GDP_Per_Capita' to numeric, coercing errors
df_gdp_melted['Year'] = pd.to_numeric(df_gdp_melted['Year'], errors='coerce').astype(int)
df_gdp_melted['GDP_Per_Capita'] = pd.to_numeric(df_gdp_melted['GDP_Per_Capita'], errors='coerce')

# Rename 'GeoName' to 'State'
df_gdp_cleaned = df_gdp_melted.rename(columns={'GeoName': 'State'})

print("Head of Cleaned and Melted GDP Data:")
display(df_gdp_cleaned.head())

print("\nInfo of Cleaned and Melted GDP Data:")
df_gdp_cleaned.info()
```

Рассчитайте среднее значение `Average_Cigarette_Price` для каждого `State` за эти годы, сохранив результат в DataFrame `df_average_cigarette_price`.


**Reasoning**:
The instructions require filtering the raw tobacco data, selecting and renaming columns, cleaning the price column, filtering by year, and then calculating the average price per state. All these steps are related to preparing the `df_average_cigarette_price` DataFrame.

```python
import numpy as np

# Filter for columns that contain percentage data (excluding the 'Number' columns)
percentage_cols = [col for col in df_age_raw.columns if '!!Percent' in col]
df_age_percentages = df_age_raw[['Label (Grouping)'] + percentage_cols].copy()

# Clean up column names to just state names
df_age_percentages.columns = ['Label (Grouping)'] + [col.split('!!')[0] for col in percentage_cols]

# Clean up 'Label (Grouping)' values by stripping whitespace
df_age_percentages['Label (Grouping)'] = df_age_percentages['Label (Grouping)'].str.strip()

print("Head of Age Demographics Data (Percentages Only):")
display(df_age_percentages.head())
print("\nInfo of Age Demographics Data (Percentages Only):")
df_age_percentages.info()
```

```python
# Extract rows for '15 to 19 years' and '20 to 24 years'
df_15_19 = df_age_percentages[df_age_percentages['Label (Grouping)'] == '15 to 19 years'].drop(columns=['Label (Grouping)']).iloc[0]
df_20_24 = df_age_percentages[df_age_percentages['Label (Grouping)'] == '20 to 24 years'].drop(columns=['Label (Grouping)']).iloc[0]

# Sum the percentages for '15 to 19 years' and '20 to 24 years' for each state
df_combined_age_percentages = pd.DataFrame({
    'State': df_15_19.index,
    'Percentage_15_24': df_15_19.astype(float) + df_20_24.astype(float)
}).reset_index(drop=True)

non_state_entries = ['United States', 'Puerto Rico']
df_age_demographics_cleaned = df_combined_age_percentages[~df_combined_age_percentages['State'].isin(non_state_entries)].copy()

# Ensure 'State' column is clean (e.g., remove leading/trailing spaces if any)
df_age_demographics_cleaned['State'] = df_age_demographics_cleaned['State'].str.strip()

print("Head of Cleaned Age Demographics Data (15-24%):")
display(df_age_demographics_cleaned.head())
print("\nInfo of Cleaned Age Demographics Data (15-24%):")
df_age_demographics_cleaned.info()
```

Обновите процесс объединения, чтобы использовать `df_tobacco_consumption_yearly` вместо `df_average_tobacco_consumption`.

```python
from functools import reduce

data_frames = [
    df_average_beer_consumption,
    df_average_gdp,
    df_tobacco_consumption_yearly,  # Changed from df_average_tobacco_consumption
    df_average_cigarette_price,
    df_age_demographics_cleaned
]

df_merged_final = reduce(lambda left, right: pd.merge(left, right, on='State', how='inner'), data_frames)

# List of states to exclude
excluded_states = [
    'Alaska', 'Hawaii', 'Maryland', 'Michigan', 'New Jersey',
    'New York', 'Washington', 'Massachusetts', 'Arizona', 'Oregon', 'Florida'
]

# Filter out the excluded states
df_merged_final = df_merged_final[~df_merged_final['State'].isin(excluded_states)].copy()

print("Head of Final Merged DataFrame (Excluded States):")
display(df_merged_final.head())

print("\nInfo of Final Merged DataFrame (Excluded States):")
df_merged_final.info()
```

```python
df_cigarette_consumption_all_years = df_tobacco_raw[
    df_tobacco_raw['SubMeasureDesc'] == 'Cigarette Consumption (Pack Sales Per Capita)'
].copy()

df_cigarette_consumption_all_years = df_cigarette_consumption_all_years[[
    'LocationDesc', 'Year', 'Data_Value'
]]

df_cigarette_consumption_all_years.rename(columns={
    'LocationDesc': 'State',
    'Year': 'Year',
    'Data_Value': 'Cigarette_Packs_Per_Capita'
}, inplace=True)

# Clean 'Cigarette_Packs_Per_Capita' column: remove commas and convert to numeric
df_cigarette_consumption_all_years['Cigarette_Packs_Per_Capita'] = df_cigarette_consumption_all_years['Cigarette_Packs_Per_Capita'].str.replace(',', '.', regex=False)
df_cigarette_consumption_all_years['Cigarette_Packs_Per_Capita'] = pd.to_numeric(df_cigarette_consumption_all_years['Cigarette_Packs_Per_Capita'], errors='coerce')

# Filter for years 1970 to 2004
df_filtered_by_years = df_cigarette_consumption_all_years[
    (df_cigarette_consumption_all_years['Year'] >= 1970) &
    (df_cigarette_consumption_all_years['Year'] <= 2004)
].copy()

# List of states to exclude
excluded_states = [
    'Alaska', 'Hawaii', 'Maryland', 'Michigan', 'New Jersey',
    'New York', 'Washington', 'Massachusetts', 'Arizona', 'Oregon',
    'Florida', 'District of Columbia'
]

# Filter out the excluded states
df_cigarette_consumption_39_states = df_filtered_by_years[~df_filtered_by_years['State'].isin(excluded_states)].copy()

print("Head of Cigarette Packs Per Capita for 39 States (1970-2004):")
display(df_cigarette_consumption_39_states.head())

print("\nInfo of Cigarette Packs Per Capita for 39 States (1970-2004):")
df_cigarette_consumption_39_states.info()

# Define the output path for the new CSV file
output_csv_path_39_states = os.path.join(data_folder_path, 'Cigarette_packs_per_capita_39_states_1970_2004.csv')

# Save the DataFrame to CSV
df_cigarette_consumption_39_states.to_csv(output_csv_path_39_states, index=False)

print(f"\nDataFrame успешно сохранен в: {output_csv_path_39_states}")
```
