```python
# Загрузка данных о потреблении сигарет
primary_outcomes_df = pd.read_csv('/content/drive/MyDrive/ColabNotebooks/SCMCourseWork/dataFiles/99prop/Cigarette_packs_per_capita_39_states_1970_2004.csv')

print("Информация о загруженном датафрейме с потреблением сигарет:")
primary_outcomes_df.info()
display(primary_outcomes_df.head())
```

```python
primary_outcomes_df = pd.pivot_table(
    data=primary_outcomes_df,
    values='Cigarette_Packs_Per_Capita',
    index='Year',
    columns='State'
)
primary_outcomes_df.info()
display(primary_outcomes_df.head())
```

```python
def plot(synth_tables):
    synth_Y1, Y1 = synth_tables[0], synth_tables[1]
    plt.plot(range(len(synth_Y1)),synth_Y1, 'r--', label="Synthetic Control")
    plt.plot(range(len(synth_Y1)),Y1, 'b-', label="Actual Data")
    plt.title("Example Synthetic Control Model")
    plt.ylabel("Y axis")
    plt.xlabel("X axis")
    plt.legend(loc='upper left')
    plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import fmin_slsqp, minimize
from matplotlib import pyplot as plt

# -----------------------
import sys
sys.path.append('/content/drive/MyDrive/ColabNotebooks/SCMCourseWork/src/scmlib')

import scmsolver
from scmsolver import scm_solve
from scmsolver import basic_dataprep

X0, X1, Y0_pre, Y1_pre, Y0, Y1 = scmsolver.basic_dataprep(primary_predictors_df,
                             primary_outcomes_df,
                             'California',
                             primary_predictors_df.columns.drop('California').tolist(),
                             list(range(1970, 1989)), list(range(1970, 2005)))
# synth_tables = synth_Y1, Y1, matrix_V, vector_W, predictors_table, outcomes_table

res = scmsolver.scm_solve(Y0.values, Y0_pre.values, Y1_pre.values, X0.values, X1.values) # returned: {V, W, Y_1_synth}

# synth_tables = synth_Y1, Y1, matrix_V, vector_W, predictors_table, outcomes_table
cigarette_table = [res['Y_1_synth'], Y1, res['V'], res['W'], primary_predictors_df, primary_outcomes_df]

plot(cigarette_table)
```
